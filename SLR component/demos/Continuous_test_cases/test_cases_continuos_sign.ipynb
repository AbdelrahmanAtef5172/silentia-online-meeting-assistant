```python
import sys, os, cv2
sys.path.insert(0, os.path.abspath(".."))

from inference import ContinuousSignRecognizer
from IPython.display import Video, display

CHECKPOINT = "../best_model.pt"
MODELS_DIR = "../models"

test_videos = [
    ("Concatenated_2Videos_test_case1.mp4", "drink book"),
    ("concatenated_7videos_test_case_2.mp4", "book drink chair go before book computer"),
]
```

```python
for filename, gt in test_videos:
    path = os.path.abspath(filename)
    words = []
    sentences = []

    def on_word(w, c):
        words.append(w)

    def on_sentence(s):
        sentences.append(s)

    recognizer = ContinuousSignRecognizer(
        checkpoint_path=CHECKPOINT,
        models_dir=MODELS_DIR,
        frame_step=2,
        on_word_recognized=on_word,
        on_sentence_complete=on_sentence,
    )
    if not recognizer.ready:
        raise RuntimeError(recognizer.init_error)

    print("=" * 60)
    print(f"Video: {filename}")
    display(Video(path, embed=True, width=320))

    cap = cv2.VideoCapture(path)
    while True:
        ok, frame = cap.read()
        if not ok:
            break
        recognizer.process_frame(frame)
    cap.release()
    recognizer.flush()
    recognizer.close()

    print(f"Ground truth: {gt}")
    print(f"Recognized words: {words}")
    print(f"Sentences: {sentences}")
    all_found = all(w in words for w in gt.split())
    print(f"All words correct: {all_found}")
    print()
```
